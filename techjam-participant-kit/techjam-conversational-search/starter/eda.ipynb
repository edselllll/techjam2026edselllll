```python
import json
import pandas as pd

rows = []
with open("data/catalog.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            rows.append(json.loads(line))

# Automatically flattens nested dicts into dot-notation columns
df = pd.json_normalize(rows)
print(df.columns.tolist())
```

```python
import json
import pandas as pd

rows = []
with open("data/public_set.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            rows.append(json.loads(line))

# Automatically flattens nested dicts into dot-notation columns
df = pd.json_normalize(rows)
print(df.columns.tolist())
```

```python
# Rewind the open file so we can parse it from the start

unique_preference_tags = set()
with open("data/public_set.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            row = json.loads(line)
            tags = row.get("user_profile", {}).get("preference_tags", [])
            if isinstance(tags, list):
                unique_preference_tags.update(tags)

unique_preference_tags
```
